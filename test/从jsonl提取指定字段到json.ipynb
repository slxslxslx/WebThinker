```python
import json

# 输入文件路径
input_path = "/home/shaolingxuan/project/WebThinker/outputs/seal0.qwen3.5-9b.webthinker/all.jsonl"
# 输出文件路径
output_path = "/home/shaolingxuan/project/WebThinker/outputs/seal0.qwen3.5-9b.webthinker/all_extracted.json"

# 需要提取的字段
fields_to_extract = ["id", "Question", "answer", "Output" ]

results = []

with open(input_path, "r", encoding="utf-8") as f:
    for line_num, line in enumerate(f, 1):
        line = line.strip()
        if not line:
            continue
        try:
            data = json.loads(line)
            # 提取指定字段
            extracted = {}
            for field in fields_to_extract:
                if field in data:
                    extracted[field] = data[field]
                else:
                    print(f"警告: 第 {line_num} 行缺少字段 '{field}'")
                    extracted[field] = None
            results.append(extracted)
        except json.JSONDecodeError as e:
            print(f"错误: 第 {line_num} 行 JSON 解析失败 - {e}")

with open(output_path, "w", encoding="utf-8") as f:
    json.dump(results, f, ensure_ascii=False, indent=2)

print(f"完成！共处理 {len(results)} 条数据，结果已写入: {output_path}")
```
