```python
import json
import re

# ========== 配置路径 ==========
input_path  = "/home/shaolingxuan/project/WebThinker/outputs/seal0.qwen3.5-9b.webthinker/dev.8.3,19:50.77.json"
output_path = "/home/shaolingxuan/project/WebThinker/outputs/seal0.qwen3.5-9b.webthinker/all_extracted.json"

# ========== 1. 读取已有的 all_extracted.json ==========
try:
    with open(output_path, "r", encoding="utf-8") as f:
        all_data = json.load(f)
    print(f"读取已有数据 {len(all_data)} 条")
except (FileNotFoundError, json.JSONDecodeError):
    all_data = []
    print("输出文件不存在或为空，从头开始")

# ========== 2. 读取输入 JSON 文件 ==========
with open(input_path, "r", encoding="utf-8") as f:
    raw_data = json.load(f)

# 如果是单条 dict，包装成 list
if isinstance(raw_data, dict):
    raw_data = [raw_data]

print(f"输入文件共 {len(raw_data)} 条数据")

# ========== 3. 正则匹配 \boxed{...} ==========
pattern = r'\\boxed\{(.*)\}'    # ← 用非贪婪 .*?，见下方说明。保持和webthinker一致，就用贪婪模式

matched_count   = 0
unmatched_count = 0

for item in raw_data:
    # 兼容 "Output" 和 "output" 两种字段名
    output = item.get("Output", item.get("output"))
    if output is None:
        print(f"  [跳过] id={item.get('id')} 无 Output 字段")
        unmatched_count += 1
        continue

    matches = re.findall(pattern, output)
    if matches:
        pred_answer = matches[-1]  # 取最后一个匹配
        extracted = {
            "id":           item.get("id"),
            "Question":     item.get("Question"),
            "answer":       item.get("answer"),
            "Output":       output,
            "Pred_Answer":  pred_answer
        }
        all_data.append(extracted)
        matched_count += 1
        print(f"  [匹配] id={item.get('id')}  Pred_Answer={pred_answer}")
    else:
        unmatched_count += 1
        print(f"  [未匹配] id={item.get('id')}  Output 中无 \\boxed{{}}")

# ========== 4. 写回 all_extracted.json ==========
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(all_data, f, ensure_ascii=False, indent=2)

print(f"\n完成！匹配成功 {matched_count} 条，未匹配 {unmatched_count} 条")
print(f"输出文件现有 {len(all_data)} 条数据，路径: {output_path}")
```

```python
import json
import re

# ========== 配置路径 ==========
input_path  = "/home/shaolingxuan/project/WebThinker/outputs/XbenchDS.qwen3.5-9b.webthinker/dev.8.2,17:43.88.json"
output_path = "/home/shaolingxuan/project/WebThinker/outputs/XbenchDS.qwen3.5-9b.webthinker/dev.7.31,17:29.99-all.json"

# ========== 1. 读取已有的 all_extracted.json ==========
try:
    with open(output_path, "r", encoding="utf-8") as f:
        all_data = json.load(f)
    print(f"读取已有数据 {len(all_data)} 条")
except (FileNotFoundError, json.JSONDecodeError):
    all_data = []
    print("输出文件不存在或为空，从头开始")

# ========== 2. 读取输入 JSON 文件 ==========
with open(input_path, "r", encoding="utf-8") as f:
    raw_data = json.load(f)

# 如果是单条 dict，包装成 list
if isinstance(raw_data, dict):
    raw_data = [raw_data]

print(f"输入文件共 {len(raw_data)} 条数据")

# ========== 3. 正则匹配 \boxed{...} ==========
pattern = r'\\boxed\{(.*)\}'    # ← 用非贪婪 .*?，见下方说明。保持和webthinker一致，就用贪婪模式

matched_count   = 0
unmatched_count = 0

for item in raw_data:
    # 兼容 "Output" 和 "output" 两种字段名
    output = item.get("Output", item.get("output"))
    if output is None:
        print(f"  [跳过] id={item.get('id')} 无 Output 字段")
        unmatched_count += 1
        continue

    matches = re.findall(pattern, output)
    if matches:
        pred_answer = matches[-1]  # 取最后一个匹配
        extracted = {
            "id":           item.get("id"),
            "Question":     item.get("Question"),
            "answer":       item.get("answer"),
            "Output":       output,
            "Pred_Answer":  pred_answer
        }
        all_data.append(extracted)
        matched_count += 1
        print(f"  [匹配] id={item.get('id')}  Pred_Answer={pred_answer}")
    else:
        unmatched_count += 1
        print(f"  [未匹配] id={item.get('id')}  Output 中无 \\boxed{{}}")

# ========== 4. 写回 all_extracted.json ==========
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(all_data, f, ensure_ascii=False, indent=2)

print(f"\n完成！匹配成功 {matched_count} 条，未匹配 {unmatched_count} 条")
print(f"输出文件现有 {len(all_data)} 条数据，路径: {output_path}")
```
